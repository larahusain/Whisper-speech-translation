```python
import sys
import pathlib
import pandas as pd
```

```python
# path_en = "dev_en.tsv"
# path_yue = "dev_yue.tsv"
# path_output  = "dev_yue_en.tsv"

# path_en = "test_en.tsv"
# path_yue = "test_yue.tsv"
# path_output  = "test_yue_en.tsv"

path_en = "train_en.tsv"
path_yue = "train_yue.tsv"
path_output  = "train_yue_en.tsv"

df_en = pd.read_csv(path_en,sep="\t",header=None,
                    usecols=[0, 2],
                    names=["id", "text_en"],
                    dtype=str,
                    keep_default_na=False,
    )
df_yue = pd.read_csv(path_yue,sep="\t",header=None,
                     usecols=[0, 1, 2],
                    names=["id", "audio_path", "text_source"],
                    dtype=str,
                    keep_default_na=False,
    )
```

```python
df_yue
```

```python
# eng_df and cant_df may contain multiple entries for one id
# so keep only one english tranlation
# but keep all cantonese because it has different version of audios

df_en_unique  = df_en.drop_duplicates(subset="id", keep="first")
```

```python
merged = pd.merge(df_yue, df_en_unique,on="id",how="left", suffixes=("_source", "_en"))
```

```python
merged.to_csv(path_output,sep="\t",index=False,header=True)
```
